## Adding Memory In Agentic Graph
- memory is stored in your application's RAM, not in a database or on disk.

#### Base

```python
from typing_extensions import TypedDict

from langgraph.graph import StateGraph,START, END
from langgraph.graph.message import add_messages

import os
from dotenv import load_dotenv
load_dotenv()
```

```python
from typing import Annotated
class State(TypedDict):
    messages:Annotated[list, add_messages]
```

```python
from langchain.chat_models import init_chat_model

# llm = init_chat_model("google_genai:gemini-2.5-flash-lite")
llm = init_chat_model("groq:qwen/qwen3.8-27b")
llm.profile
```

#### StateGraph

```python
## Stategraph
from langgraph.graph import StateGraph, START, END
from langgraph.prebuilt import ToolNode
from langgraph.checkpoint.memory import MemorySaver
```

```python
memory = MemorySaver()  # initalize the memory

builder = StateGraph(State)
```

```python
def llm_call(state:State):
    return{ "messages": [llm.invoke(state["messages"])]}
```

```python
builder.add_node("calling_llm", llm_call)

builder.add_edge(START, "calling_llm")
builder.add_edge("calling_llm",END)

graph = builder.compile(checkpointer=memory) # add the memory in graph
```

```python
config={"configurable":{"thread_id":"1"}} # to track the memory of a thread
```

### Check the memory

```python
response=graph.invoke({"messages":"Hello, My name is Naveen"},config=config)
for m in response['messages' ]:
    m.pretty_print()
```
